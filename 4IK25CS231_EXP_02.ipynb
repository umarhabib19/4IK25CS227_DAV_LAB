```python
import pandas as pd
import numpy as np

data={
    'Name': ['Amit','Raju','Chetan','Amit','Divya','Ravi','Karan','Karan',
             'Pooja','Manish','Anjali','Suresh','Kavya','Rahul','Meera',
             'Vikram','Nisha','Arjun','Divya','Rohit','Swathi','Naveen',
            'Priya','Karan','Deepa'],
    'Age': [25,np.nan,30,25,28,22,24,np.nan,27,26,23,29,np.nan,
            25,22,28,24,np.nan,28,26,23,30,27,np.nan,25],
    'Marks': [85,90,np.nan,85,78,95,88,72,np.nan,91,67,84,79,
              np.nan,93,88,76,82,np.nan,np.nan,90,85,71,72,89],
    'City': ['Mysuru','Mysuru','Bengaluru','Mysuru',np.nan,'Delhi','Mysuru',
             'Manglore','Bengaluru',np.nan,'Mysuru','Hubli','Manglore',
             'Delhi','Bengaluru',np.nan,'Mysuru','Hubli',np.nan,'Delhi',
             'Manglore','Mysuru','Bengaluru','Manglore','Hubli']
}
df = pd.DataFrame(data)
print("Dataset:")
print(df)
```

```python
print("\nShape of data set(rows,columns):",df.shape)
print("\n Columns names:",df.columns.tolist())
print("\nData type of each element:")
print(df.dtypes)
```

```python
print("\nData tpes of each column:")
print(df.dtypes)
print("\nDataset info:")
df.info()
print("\nDescriptive statistics (numericalcolumns):")
print(df.describe())
```

```python
print("\nMean:\n",df.mean(numeric_only=True))
print("\nMedian:\n",df.median(numeric_only=True))
print("\nStandard Deviation:\n",df.std(numeric_only=True))
```

```python
print("\nMissing values in each coloumns:")
print(df.isnull().sum())
print("\nTotal missing values in data set:",df.isnull().sum().sum())
print("\nDuplicate rows in dataset:")
print(df[df.duplicated()])
print("\nNumber of duplicates rows:",df.duplicated().sum())
print("\nEDA process is completed successfully!")
```
