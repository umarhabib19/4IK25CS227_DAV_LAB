```python
# SWIGGY BANGALORE DATA ANALYSIS

# 1. Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


# 2. Load Dataset
df = pd.read_csv('/content/Swiggy Bangalore.csv')

print("Dataset Loaded Successfully!")


# 3. Data Exploration
print("\nFirst Five Records:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())


# 4. Data Cleaning
df = df.drop(['Offer', 'URL'], axis=1)

df = df.drop_duplicates()

df['Rating'] = pd.to_numeric(df['Rating'], errors='coerce')

df['Cost for Two (in Rupees)'] = pd.to_numeric(
    df['Cost for Two (in Rupees)'], errors='coerce'
)

df = df.dropna(subset=['Cost for Two (in Rupees)', 'Area'])

print("\nData Cleaning Completed!")
print("Cleaned Dataset Shape:", df.shape)


# 5. Data Preprocessing
rating_data = df.dropna(subset=['Rating'])

print("\nRecords with Available Ratings:", len(rating_data))


# 6. Exploratory Data Analysis
print("\nStatistical Summary:")
print(df.describe())

print("\nAverage Restaurant Rating:",
      round(rating_data['Rating'].mean(), 2))

print("\nAverage Cost for Two:",
      round(df['Cost for Two (in Rupees)'].mean(), 2))


# 7. Data Visualization

# Graph 1: Restaurants by Area
plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='Area',
              order=df['Area'].value_counts().head(10).index)
plt.title("Top 10 Areas with Most Restaurants")
plt.xlabel("Area")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# Graph 2: Rating Distribution
plt.figure(figsize=(8, 5))
sns.histplot(data=rating_data, x='Rating', bins=10)
plt.title("Restaurant Rating Distribution")
plt.xlabel("Rating")
plt.ylabel("Number of Restaurants")
plt.tight_layout()
plt.show()


# Graph 3: Cost Distribution
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x='Cost for Two (in Rupees)', bins=15)
plt.title("Restaurant Cost Distribution")
plt.xlabel("Cost for Two (Rs.)")
plt.ylabel("Number of Restaurants")
plt.tight_layout()
plt.show()


# 8. Generate Insights
print("\nProject Insights")

print("\nMost Common Area:")
print(df['Area'].value_counts().idxmax())

print("\nHighest Rated Restaurant:")
print(rating_data.loc[
    rating_data['Rating'].idxmax(), 'Restaurant Name'
])

print("\nAverage Restaurant Rating:")
print(round(rating_data['Rating'].mean(), 2))

print("\nAverage Cost for Two:")
print(round(df['Cost for Two (in Rupees)'].mean(), 2))


# 9. Conclusion
print("\nConclusion")

print("""
The Swiggy Bangalore dataset was analysed using Python.

Data cleaning and preprocessing were performed.
Exploratory data analysis was carried out.
Seaborn was used to create different visualizations.

The analysis helped us understand restaurant
distribution, ratings and pricing patterns in Bangalore.
""")
```
